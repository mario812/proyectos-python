```python
from PIL import Image
import pytesseract

# Ruta a la imagen
imagen = "musical/Screenshot_2026-01-16-20-26-02-396_com.facebook.katana.jpg"

# Cargar imagen
img = Image.open(imagen)

# Extraer texto con OCR
#exto = pytesseract.image_to_string(img, lang="spa")  # usa español si los nombres tienen acentos
texto = pytesseract.image_to_string(img)

# Guardar en archivo de texto
ruta = "/musical/texto-1968/texto_1.txt"
with open(ruta, "w", encoding="utf-8") as f:
    f.write(texto)

print("Proceso completado. Archivo generado: nombres_extraidos.txt")
```
