# Anotación de Datos
Algunas veces conviene colocarle descripciones a los campos del esquema. Esto se logra con la dependencia Annotated que debe importarse

```python
from pydantic import BaseModel, EmailStr, AnyUrl, Field
from pydantic import ValidationError
from typing import List, Dict, Optional, Annotated
```

```python
class PatientData(BaseModel):
    name: Annotated[str, Field(max_length=50, title='nombre del paciente',
                description='El nombre del paciente no puede ser mayot de 50 caracteres',
                example=['Pedro Perez', 'Juan Cabeza'])]
    email: EmailStr
    linkedin_url: Optional[AnyUrl] = None
    age: int
    weight: Annotated[float, Field(gt=0, description='El peso debe estar en kg')]
    married: Annotated[bool, Field(default=None, description='Indicar si el paciente es casado o no')]
    allergies: Annotated[Optional[list[str]],Field(default=None,max_length=5)]
    contact_info: dict[str, str]

def add_patient_data(patient: PatientData):
        print(patient.name)
        print(patient.email)
        print(patient.linkedin_url)
        print(patient.age)
        print(patient.weight)
        print(patient.allergies)
        print(patient.married)
        print("Los datos se agregaron correctamente a la base de datos")

```

```python
# modificar el esquema para agregar un nuevo campo email
patient_data = {"name": "Bappy", "email": "bappy@example.com",
                "linkedin_url": "https:linkedin.com/in/bappy",
                "age": "25", "weight": "70.5",
                "allergies": ["pollen", "gluten", "dust", "penicilline"],
                "contact_info": {"phone": "123-456-7890" }}
patient_1= PatientData(**patient_data)
add_patient_data(patient_1)



```

A continuación se lanza un ejemplo un poco más complicado, en el cual la aplicación ve a validar el paciente para establecer si es candidato a un tipo de beneficio.

```python
from pydantic import BaseModel, EmailStr, AnyUrl, Field, field_validator
from typing import List, Dict, Optional, Annotated

class PatientData(BaseModel):
    name: str
    email: EmailStr
    linkedin_url: AnyUrl
    age: int
    weight: float
    married: bool
    allergies: List[str]
    contact_info: dict[str, str]

    @field_validator('email')
    def email_validator(cls, value):
        valid_domains = ['hdfc.com', 'icici.com']
        domain_name = value.split('@')[-1]
        if domain_name not in valid_domains:
            raise ValueError("Invalid email domain")
        return value
  
def add_patient_data(patient: PatientData):
        print(patient.name)
        print(patient.email)
        print(patient.linkedin_url)
        print(patient.age)
        print(patient.weight)
        print(patient.allergies)
        print(patient.married)
        print("Los datos se agregaron correctamente a la base de datos")
 

```

```python
patient_data = {"name": "Bappy", "email": "bappy@hdfc.com",
                "linkedin_url": "https:linkedin.com/in/bappy",
                "age": "25", "weight": "70.5",
                "married": "True",
                "allergies": ["pollen", "gluten", "dust", "penicilline"],
                "contact_info": {"phone": "123-456-7890" }}
patient_1= PatientData(**patient_data)
add_patient_data(patient_1)
```
