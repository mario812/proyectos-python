```python
from pydantic import BaseModel
from pydantic import ValidationError
```

```python
class PatientData(BaseModel):
    name: str
    age: int
    weight: float


def add_patient_data(patient: PatientData):
    print(patient.name)
    print(patient.age)
    print(patient.weight)
    
    print("Los datos se agregaron correctamente a la base de datos")
```

```python
from pydantic import ValidationError

# Pydantic convierte una cadena numerica a entero
patient_2_data = {"name": "Maria Garcia", "age": "30"}
patient_2 = PatientData(**patient_2_data)
print("Edad convertida:", patient_2.age, type(patient_2.age).__name__)
add_patient_data(patient_2)

# Una palabra como "treinta" no se puede convertir automaticamente
incorrect_patient_data = {"name": "Maria Garcia", "age": "treinta"}
try:
    PatientData(**incorrect_patient_data)
except ValidationError as error:
    print("Dato rechazado:", error.errors()[0]["msg"])

print("La ejecucion continua")
```
