```python
from PIL import Image
import pytesseract

# Ruta a la imagen
imagen = "/musical/1968/Screenshot_2026-01-16-20-26-02-396_com.facebook.katana.jpg"

# Cargar imagen
img = Image.open(imagen)
```

```python
# Extraer texto con OCR
#texto = pytesseract.image_to_string(img, lang="spa")  # usa español si los nombres tienen acentos
texto = pytesseract.image_to_string(img)

# Guardar en archivo de texto
with open("nombres_extraidos.txt", "w", encoding="utf-8") as f:
    f.write(texto)

print("Proceso completado. Archivo generado: nombres_extraidos.txt")
```
