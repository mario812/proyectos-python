```python
from pydantic import BaseModel
from pydantic import ValidationError
from typing import List, Dict
```

```python
class PatientData(BaseModel):
    name: str
    age: int


def add_patient_data(patient: PatientData):
    print(patient.name)
    print(patient.age)

    print("Los datos se agregaron correctamente a la base de datos")

def carga_datos(paciente):
    try:
        patient = PatientData(**paciente)
        return patient
    except ValidationError as error:
         print("Dato rechazado:", error.errors()[0]["msg"])
         return None
```

# Pydantic
Para usar la validación pedántica de datos se deben tener en cuenta los siguientes aspectos:
1. Definir un modelo pedántico que tome en cuenta los tipos de datos presentes en el esquema de desarrollo. Aquí se incluyen los campos que se van a usar y cualquier condición de validación (por ejemplo > 0).
2. Instanciar el modelo con datos de entrada sin formato usando un diccionario o un archivo JSON. El modelo pedántico validará automáticamente los datos forzandolos a asumir el tipo correcto en Python (si es posible). Si los datos no cumplen con los requisitos del modelo, Pydantic levanta un mensaje ValidationError.
3. Pasarle el objeto del modelo validado a las funciones o usarlo en todo el código. Esto asegura que cada sección del programa opera con datos limpios, con tipo seguro y lógicamente válidos.


```python
#los datos se ingresan mediante diccionario o archivo JSON
# se ingresan datos correctos
patient_1_data = {"name":"Juan Pérez", "age":30} #diccionario con los datos
patient_1 = carga_datos(patient_1_data)
```

```python
add_patient_data(patient_1)
```

```python
# se ingresan datos incorrectos, pero pydantic corrige la cadena en age a entero
patient_2_data = {"name": "Maria Garcia", "age": "30"}
patient_2 = carga_datos(patient_1_data)
```

```python
add_patient_data(patient_2)
```

```python
# se ingresan datos incorrectos, pydantic no puede corregir la cadena en age a entero
patient_3_data = {"name": "Carlos Lopez", "age": "treinta"}
patient_3 = carga_datos(patient_3_data)
```
