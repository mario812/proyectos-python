```python
import os
import re
from fuzzywuzzy import fuzz

# Carpeta donde están los archivos .txt
carpeta = "musical/1968/texto-1968"

# Archivo de salida
salida = "/musical/1968/texto-1968/canciones_unicas.txt"

# Normalización básica para títulos y autores
def normalizar(texto):
    texto = texto.lower()
    texto = texto.replace("|", "")
    texto = texto.replace("  ", " ")
    texto = texto.replace("–", "-")
    texto = texto.replace("—", "-")
    texto = texto.replace("“", "").replace("”", "")
    texto = texto.strip()
    return texto

# Extraer título y autor del formato "titulo - autor"
def extraer(linea):
    if "-" not in linea:
        return None
    partes = linea.split("-", 1)
    titulo = normalizar(partes[0].strip())
    autor = normalizar(partes[1].strip())
    return titulo, autor

# Lista final de canciones únicas
canciones_unicas = []

# Recorrer todos los archivos de la carpeta
for archivo in os.listdir(carpeta):
    if archivo.endswith(".txt"):
        ruta = os.path.join(carpeta, archivo)
        with open(ruta, "r", encoding="utf-8") as f:
            for linea in f:
                linea = linea.strip()
                datos = extraer(linea)
                if not datos:
                    continue

                titulo, autor = datos
                cancion = f"{titulo} - {autor}"

                # Verificar si ya existe una canción equivalente
                duplicado = False
                for existente in canciones_unicas:
                    titulo_exist, autor_exist = existente.split(" - ")

                    # Comparación difusa de títulos y autores
                    if fuzz.ratio(titulo, titulo_exist) > 90 and fuzz.ratio(autor, autor_exist) > 90:
                        duplicado = True
                        break

                if not duplicado:
                    canciones_unicas.append(cancion)

# Ordenar alfabéticamente
canciones_unicas = sorted(canciones_unicas)

# Guardar archivo final
with open(salida, "w", encoding="utf-8") as f:
    for c in canciones_unicas:
        f.write(c + "\n")

print("Archivo generado:", salida)
```
